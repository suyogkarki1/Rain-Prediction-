```python
X=df.drop(columns=["rain","precipitation"])
y=df["rain"]
```

```python

print(X.shape)
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=42,test_size=0.2)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import (RandomForestClassifier, GradientBoostingClassifier, 
                              AdaBoostClassifier, ExtraTreesClassifier)
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score, 
                           roc_auc_score, roc_curve, confusion_matrix, classification_report,
                           mean_squared_error, mean_absolute_error, r2_score)
from sklearn.model_selection import train_test_split, KFold ,cross_val_score
```

```python
models={'Logistic Regression':LogisticRegression(max_iter=100),
    'Decision Tree': DecisionTreeClassifier(random_state=42),
    'Random Forest': RandomForestClassifier(n_estimators=100,random_state=42),
    'KNN':KNeighborsClassifier(n_neighbors=6)}

results=[]

for model in models.values():
    kf = KFold(n_splits=6, random_state=42, shuffle=True)
    cv_results = cross_val_score(model, X_train_scaled, y_train, cv=kf)
    results.append(cv_results)

plt.boxplot(results,labels= models.keys())
plt.show()
```

Random Forest achieved the best performance (~0.86–0.88), followed by Logistic Regression (~0.84–0.85), KNN (~0.83–0.84), while Decision Tree had the lowest accuracy (~0.78–0.80).

```python
for name,model in models.items():
    model.fit(X_train_scaled,y_train)
    test_score=model.score(X_test_scaled,y_test)
    print("{} Test Set Accuracy:{}".format(name,test_score))
```

```python
for name,model in models.items():
    model.fit(X_train_scaled,y_train)
    y_pred=model.predict(X_test_scaled)
    y_pred_proba = model.predict_proba(X_test_scaled)[:, 1]


    #metric
    accuracy=accuracy_score(y_test,y_pred)
    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    f1= f1_score(y_test, y_pred)
    auc = roc_auc_score(y_test, y_pred_proba)



    print(f"\n{name}:")
    print(f"  Accuracy: {accuracy:.4f}")
    print(f"  F1-Score: {f1:.4f}")
    print(f"  ROC-AUC: {auc:.4f}")
```

Among all classification models, Random Forest achieved the best overall performance with an accuracy of 86.89%, F1-score of 83.12%, and ROC-AUC score of 92.88%. Although KNN achieved the same accuracy as Random Forest, Random Forest performed better in F1-score and ROC-AUC, which means it is more reliable for distinguishing between rainy and non-rainy days. Decision Tree showed the weakest performance, suggesting that a single tree may not generalize well for this dataset.

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# choose best model (example: Logistic Regression)
best_model = RandomForestClassifier(random_state=42)
best_model.fit(X_train_scaled,y_train)

# predictions
y_pred = best_model.predict(X_test_scaled)

# confusion matrix
cm = confusion_matrix(y_test, y_pred)

# display
disp = ConfusionMatrixDisplay(confusion_matrix=cm,display_labels=["No Rain","Rain"])
disp.plot()

plt.title("Confusion Matrix - Random Forest")
plt.show()
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test,y_pred, target_names=['NO Rain','Rain']))
```

```python
from sklearn.metrics import roc_curve, roc_auc_score
import matplotlib.pyplot as plt

# Best model
best_model = RandomForestClassifier(random_state=42)
best_model.fit(X_train_scaled, y_train)

# Probability of class 1: Rain
y_prob = best_model.predict_proba(X_test_scaled)[:, 1]

# ROC values
fpr, tpr, thresholds = roc_curve(y_test, y_prob)
auc_score = roc_auc_score(y_test, y_prob)

# Plot ROC curve
plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, label=f"Random Forest AUC = {auc_score:.4f}")
plt.plot([0, 1], [0, 1], linestyle="--", label="Random Guess")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - Random Forest Rain Prediction")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import pandas as pd

# Month names dictionary
month_names = {
    1: "January",
    2: "February",
    3: "March",
    4: "April",
    5: "May",
    6: "June",
    7: "July",
    8: "August",
    9: "September",
    10: "October",
    11: "November",
    12: "December"
}

# Empty list to store results
monthly_predictions = []

for m in range(1, 13):
    
    # Calculate average historical weather values for each month
    month_avg = df[df["month"] == m][X.columns].mean(numeric_only=True)
    
    # Convert average values to dataframe
    future_month = pd.DataFrame([month_avg])
    
    # Set future year and month
    future_month["year"] = 2026
    future_month["month"] = m
    
    # Keep same column order as training data
    future_month = future_month[X.columns]
    
    # Scale the future month data
    future_month_scaled = scaler.transform(future_month)
    
    # Predict rain/no rain
    prediction = best_model.predict(future_month_scaled)
    
    # Predict rain probability
    probability = best_model.predict_proba(future_month_scaled)[:, 1]
    
    # Store results
    monthly_predictions.append({
        "Year": 2026,
        "Month": m,
        "Month Name": month_names[m],
        "Rain Prediction": "Rain" if prediction[0] == 1 else "No Rain",
        "Rain Probability (%)": round(probability[0] * 100, 2)
    })

# Convert results into table
monthly_prediction_df = pd.DataFrame(monthly_predictions)

# Display final table
monthly_prediction_df
```
